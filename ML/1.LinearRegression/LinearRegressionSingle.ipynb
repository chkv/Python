```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df=pd.read_csv("homeprices.csv")
```

```python
df.head()
```

```python
%matplotlib inline
plt.xlabel("area")
plt.ylabel("USD")
plt.scatter(df.area,df.price,color="red",marker="*")
plt.show()
```

```python
reg=linear_model.LinearRegression()
reg.fit(df[['area']],df[['price']])
```

```python
reg.predict([[3122]])
```

```python
reg.predict(pd.DataFrame([[2000]],columns=['area']))
```

```python
# y=mx+b
```

```python
#m
reg.coef_
```

```python
#  b
reg.intercept_
```

```python
reg.predict(pd.DataFrame([[5000]],columns=['area']))
```

```python
reg.predict(pd.DataFrame([[1000]],columns=['area']))
```

```python
d=pd.read_csv("areas.csv")
```

```python
d.head()
```

```python
p= reg.predict(d)
```

```python
d['price']=p
```

```python
d
```

```python
d.to_csv('prediction.csv',index=False)
```

```python
%matplotlib inline
plt.xlabel("area")
plt.ylabel("USD")
plt.scatter(df.area,df.price,color="red",marker="*")
plt.plot(df.area,reg.predict(df[['area']]))
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```
