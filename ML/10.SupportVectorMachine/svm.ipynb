```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
dir(iris)
```

```python
iris.feature_names
```

```python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df.head()
```

```python
df['target']=iris.target
```

```python
df.head()
```

```python
input=df.drop('target',axis='columns')
input
```

```python
output=df[
    'target'
]
```

```python
output
```

```python
df0=df[df['target']==0]
df1=df[df['target']==1]
df2=df[df['target']==2]
```

```python
df2
```

```python
plt.scatter(df0['sepal width (cm)'],df0['sepal length (cm)'],marker='o',c='red')
plt.scatter(df1['sepal width (cm)'],df1['sepal length (cm)'],marker='x',c='blue')
plt.xlabel('sepal_width')
plt.ylabel('sepal_length')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(input,output,test_size=0.2)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn.svm import SVC
```

```python
model=SVC(C=100)
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
y_predicted=model.predict(x_test)
```
