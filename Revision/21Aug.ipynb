
6. enumerate() + zip()
----------------------
Given:

    names = ["A", "B", "C"]
    marks = [90, 85, 95]

Use:
    - enumerate() to print index + name
    - zip() to combine names + marks

```python
names = ["A", "B", "C"]
marks = [90, 85, 95]
l=list(zip(names,marks))
for x,y in enumerate(l,start=1):
    print(x,y)
```

*args
---------
Write a function:

    calculate_sum(*numbers)

that accepts any number of numbers and returns their sum.

Example:
    calculate_sum(10, 20, 30, 40)
    → 100

```python
def calc_sum(*args):
    return sum(args)

print(calc_sum(1,2,3,4,5))
```

**kwargs
-----------
Write a function:

    student_info(**details)

that prints every key and value.

Example:
    student_info(name="Vardhan", age=22, course="AI")

Output:
    name Vardhan
    age 22
    course AI

```python
def print_kwargs(**kwargs):
    for k,v in kwargs.items():
        print(k,v)




student_info={"name":"Vardhan", "age":22, "course":"AI"}
print_kwargs(**student_info)
```


8. map() + filter()
-------------------
Given:

    numbers = [1, 2, 3, 4, 5, 6]

Use:
    - map() → square numbers
    - filter() → keep even numbers


```python
numbers=[1,2,3,4,5]


sq=list(map(lambda x:x**2,numbers))


print(sq)

even=list(filter(lambda x:x%2==0,sq))
print(even)
```
